## 3. calculations.py — the maths engine
No window code here, so it can be tested alone.
An expression is cleaned (`prepare`), read as a tree (`ast`), checked against a whitelist, then evaluated.
Python's `eval()` is never used on what the user types.
Also in this file: variables, memory, history and the equation solver (SymPy).

```python
%%writefile calculations.py
"""
calculations.py - Calc's maths engine.
There is NO GUI code in this file, so it can be tested on its own (see tests.py).

How an expression becomes an answer:
    text typed by the user  ->  prepare()  ->  ast.parse()  ->  safe check  ->  evaluate
We never call Python's eval() on user input. We read the expression as a tree
(ast) and only calculate the kinds of nodes we allow.
"""
import ast
import math
import operator
import re


# ==================== SECTION 1: ERRORS AND SETTINGS ====================
class CalcError(Exception):
    """Raised with a short, friendly message the interface can show directly."""


MAX_HISTORY = 30
MAX_POWER = 10000          # stops things like 9**9**9 from freezing the app
MAX_FACTORIAL = 170        # 171! is too big for a normal decimal number

# Names a user may not use as variable names
FUNCTION_NAMES = {
    "sin", "cos", "tan", "asin", "acos", "atan",
    "sinh", "cosh", "tanh", "asinh", "acosh", "atanh",
    "ln", "log", "exp", "sqrt", "cbrt", "root", "abs", "fact", "nPr", "nCr",
}
RESERVED_NAMES = FUNCTION_NAMES | {"pi", "e", "ans"}

# "a = 5"  (a single name, then =, then a value)
ASSIGNMENT = re.compile(r"^\s*([A-Za-z]+)\s*=(?!=)\s*(.+)$")


# ==================== SECTION 2: TEXT CLEAN-UP AND NUMBER FORMAT ====================
def prepare(text):
    """Turn what the user sees (× ÷ − π ^ %) into plain Python maths."""
    s = text.strip()
    s = (s.replace("×", "*").replace("÷", "/").replace("−", "-").replace("–", "-")
          .replace("π", "pi").replace("^", "**").replace("%", "/100"))
    # implicit multiplication:  2pi -> 2*pi,  3(4+1) -> 3*(4+1),  )( -> )*(
    # (the (?!e[+-]?\d) part keeps scientific notation such as 1e5 working)
    s = re.sub(r"(?<=[\d)])(?=(?!e[+-]?\d)[A-Za-z(])", "*", s)
    # close any missing brackets, like a real calculator:  sin(30  ->  sin(30)
    s += ")" * (s.count("(") - s.count(")"))
    return s


def format_number(value):
    """Show a number with at most 12 significant digits (so 0.1+0.2 shows 0.3)."""
    if isinstance(value, int):
        if abs(value) < 10 ** 15:
            return str(value)
        value = float(value)
    if value == 0:
        return "0"                       # also removes the ugly "-0"
    return f"{value:.12g}"


# ==================== SECTION 3: SAFE PARSING ====================
_ALLOWED_NODES = (
    ast.Expression, ast.BinOp, ast.UnaryOp, ast.Constant, ast.Name, ast.Call,
    ast.Load, ast.Add, ast.Sub, ast.Mult, ast.Div, ast.Pow, ast.UAdd, ast.USub,
)


def parse_safely(prepared):
    """Parse the text and reject anything that is not plain maths."""
    try:
        tree = ast.parse(prepared, mode="eval")
    except (SyntaxError, ValueError):
        raise CalcError("Invalid expression")
    for node in ast.walk(tree):
        if not isinstance(node, _ALLOWED_NODES):
            raise CalcError("Invalid expression")
        if isinstance(node, ast.Constant) and (
                isinstance(node.value, bool) or not isinstance(node.value, (int, float))):
            raise CalcError("Invalid number")
        if isinstance(node, ast.Call) and (
                not isinstance(node.func, ast.Name) or node.keywords):
            raise CalcError("Invalid function call")
    return tree


# ==================== SECTION 4: THE CALCULATOR CLASS ====================
class Calculator:
    """Holds everything the calculator remembers: angle mode, variables,
    memory (M+), the last answer and the history."""

    _BINARY = {ast.Add: operator.add, ast.Sub: operator.sub,
               ast.Mult: operator.mul, ast.Div: operator.truediv, ast.Pow: operator.pow}

    def __init__(self):
        self.angle_mode = "DEG"      # "DEG" or "RAD"
        self.variables = {}          # named variables:  {"a": 5}
        self.memory = 0              # the M+ / M- / MR memory
        self.ans = 0                 # last answer
        self.history = []            # newest first: [(expression, result), ...]

    # ---------- angle mode ----------
    def toggle_angle_mode(self):
        self.angle_mode = "RAD" if self.angle_mode == "DEG" else "DEG"
        return self.angle_mode

    # ---------- the function table ----------
    def _functions(self):
        """All functions the user can call. Trig functions respect DEG/RAD."""
        deg = self.angle_mode == "DEG"
        to_rad = math.radians if deg else (lambda x: x)
        from_rad = math.degrees if deg else (lambda x: x)
        tidy = lambda x: round(x, 12)    # sin(180°) = 0, not 1.2e-16

        def tan(x):
            result = math.tan(to_rad(x))
            if abs(result) > 1e14:
                raise CalcError("tan is undefined at this angle")
            return tidy(result)

        def whole(x, what):
            if x != int(x) or x < 0:
                raise CalcError(f"{what} needs whole numbers that are 0 or more")
            return int(x)

        def fact(x):
            n = whole(x, "Factorial")
            if n > MAX_FACTORIAL:
                raise CalcError(f"Factorial is limited to {MAX_FACTORIAL}")
            return math.factorial(n)

        def cbrt(x):
            return tidy(math.copysign(abs(x) ** (1 / 3), x))

        def root(x, n):
            if n == 0:
                raise CalcError("Root index cannot be 0")
            if x < 0:
                if float(n).is_integer() and int(n) % 2 == 1:
                    return -tidy(abs(x) ** (1 / n))
                raise CalcError("Even root of a negative number")
            return tidy(x ** (1 / n))

        return {
            "sin": lambda x: tidy(math.sin(to_rad(x))),
            "cos": lambda x: tidy(math.cos(to_rad(x))),
            "tan": tan,
            "asin": lambda x: tidy(from_rad(math.asin(x))),
            "acos": lambda x: tidy(from_rad(math.acos(x))),
            "atan": lambda x: tidy(from_rad(math.atan(x))),
            "sinh": math.sinh, "cosh": math.cosh, "tanh": math.tanh,
            "asinh": math.asinh, "acosh": math.acosh, "atanh": math.atanh,
            "ln": math.log, "log": math.log10, "exp": math.exp,
            "sqrt": math.sqrt, "cbrt": cbrt, "root": root, "abs": abs,
            "fact": fact,
            "nPr": lambda n, r: math.perm(whole(n, "nPr"), whole(r, "nPr")),
            "nCr": lambda n, r: math.comb(whole(n, "nCr"), whole(r, "nCr")),
        }

    def _names(self):
        """Constants and variables that may appear in an expression."""
        return {"pi": math.pi, "e": math.e, "ans": self.ans, **self.variables}

    # ---------- evaluating the tree ----------
    def _eval_node(self, node, functions, names):
        if isinstance(node, ast.Expression):
            return self._eval_node(node.body, functions, names)
        if isinstance(node, ast.Constant):
            return node.value
        if isinstance(node, ast.Name):
            if node.id in names:
                return names[node.id]
            raise CalcError(f"Unknown name '{node.id}'")
        if isinstance(node, ast.UnaryOp):
            value = self._eval_node(node.operand, functions, names)
            return -value if isinstance(node.op, ast.USub) else +value
        if isinstance(node, ast.BinOp):
            left = self._eval_node(node.left, functions, names)
            right = self._eval_node(node.right, functions, names)
            if isinstance(node.op, ast.Pow) and abs(right) > MAX_POWER:
                raise CalcError("Exponent is too large")
            result = self._BINARY[type(node.op)](left, right)
            if isinstance(result, complex):
                raise CalcError("Result is not a real number")
            return result
        if isinstance(node, ast.Call):
            name = node.func.id
            if name not in functions:
                raise CalcError(f"Unknown function '{name}'")
            args = [self._eval_node(arg, functions, names) for arg in node.args]
            return functions[name](*args)
        raise CalcError("Invalid expression")

    def compute(self, text):
        """Evaluate an expression and return the number (no state is changed)."""
        tree = parse_safely(prepare(text))
        try:
            value = self._eval_node(tree, self._functions(), self._names())
            float(value)                         # raises OverflowError if far too big
        except ZeroDivisionError:
            raise CalcError("Cannot divide by zero")
        except OverflowError:
            raise CalcError("Number is too large")
        except ValueError:
            raise CalcError("Math error: input is outside the function's range")
        except TypeError:
            raise CalcError("Wrong number of values for a function")
        except RecursionError:
            raise CalcError("Expression is too deeply nested")
        if isinstance(value, float) and (math.isinf(value) or math.isnan(value)):
            raise CalcError("Number is too large")
        return value

    # ---------- the main entry point used by the interface ----------
    def evaluate(self, text):
        """Run the '=' key. Handles plain maths, assignments (a = 5) and
        equations (x + 5 = 12). Returns the text to show as the result."""
        text = text.strip()
        if not text:
            raise CalcError("Nothing to calculate")
        if "=" in text:
            match = ASSIGNMENT.match(text)
            if match:
                result = self._assign(match.group(1), match.group(2))
            else:
                result = "\n".join(self.solve(text))
        else:
            value = self.compute(text)
            self.ans = value
            result = format_number(value)
        self._add_history(text, result)
        return result

    def preview(self, text):
        """Live answer while typing. Never changes anything; returns "" if not ready."""
        text = text.strip()
        if not text or "=" in text or re.fullmatch(r"[\d.\s]+", text):
            return ""
        try:
            return format_number(self.compute(text))
        except Exception:
            return ""

    # ---------- variables ----------
    def _assign(self, name, value_text):
        if name in RESERVED_NAMES:
            raise CalcError(f"'{name}' is reserved and cannot be a variable")
        value = self.compute(value_text)
        self.variables[name] = value
        self.ans = value
        return format_number(value)

    def delete_variable(self, name):
        self.variables.pop(name, None)

    # ---------- memory (MC, MR, M+, M-) ----------
    def memory_clear(self):
        self.memory = 0

    def memory_recall(self):
        return format_number(self.memory)

    def memory_add(self, text):
        self.memory += self.compute(text)

    def memory_subtract(self, text):
        self.memory -= self.compute(text)

    # ---------- history ----------
    def _add_history(self, expression, result):
        self.history.insert(0, (expression, result.replace("\n", ";  ")))
        del self.history[MAX_HISTORY:]

    def clear_history(self):
        self.history.clear()

    # ==================== SECTION 5: EQUATION SOLVING ====================
    def solve(self, text):
        """Solve a polynomial equation with ONE unknown (degree 1 to 4).
        Returns a list of lines such as ["x = -2", "x = 2"]."""
        import sympy                              # imported here so the app starts faster
        from sympy.parsing.sympy_parser import parse_expr, standard_transformations

        if text.count("=") != 1:
            raise CalcError("An equation needs exactly one '=' sign")
        left, right = (prepare(side) for side in text.split("="))
        if not left or not right:
            raise CalcError("Both sides of '=' need a value")
        combined = f"({left}) - ({right})"       # equation becomes: something = 0
        tree = parse_safely(combined)

        # find the unknown: a name that is not a constant, variable or function
        names = self._names()
        called = {n.func.id for n in ast.walk(tree) if isinstance(n, ast.Call)}
        used = {n.id for n in ast.walk(tree) if isinstance(n, ast.Name)} - called
        unknowns = sorted(used - set(names))
        if called - {"sqrt", "cbrt", "root"}:
            raise CalcError("The solver supports polynomial equations only")
        if not unknowns:
            raise CalcError("No unknown found. Use a name with no stored value, e.g. x + 5 = 12")
        if len(unknowns) > 1:
            raise CalcError("Only one unknown is supported (found " + ", ".join(unknowns) + ")")
        for node in ast.walk(tree):
            if (isinstance(node, ast.BinOp) and isinstance(node.op, ast.Pow)
                    and isinstance(node.right, ast.Constant) and node.right.value > 10):
                raise CalcError("Exponent is too large for the solver")

        unknown = unknowns[0]
        symbol = sympy.Symbol(unknown)
        local = {name: sympy.nsimplify(value, rational=True) for name, value in names.items()}
        local.update({"pi": sympy.pi, "e": sympy.E, unknown: symbol,
                      "sqrt": sympy.sqrt, "cbrt": sympy.cbrt, "root": sympy.root})
        try:
            expr = parse_expr(combined, local_dict=local, transformations=standard_transformations)
            expr = sympy.expand(sympy.nsimplify(expr, rational=True))
            poly = sympy.Poly(expr, symbol)
        except sympy.PolynomialError:
            raise CalcError("Only polynomial equations (like x**2 - 4 = 0) are supported")
        except Exception:
            raise CalcError("Could not understand this equation")

        if poly.is_zero:
            return ["Every value works (infinitely many solutions)"]
        if poly.degree() == 0:
            return ["No solution"]
        if poly.degree() > 4:
            raise CalcError("The solver supports up to degree 4")

        real_roots = []
        for root in sympy.solve(poly.as_expr(), symbol):
            approx = sympy.N(root, 15)
            if abs(sympy.im(approx)) < 1e-9:      # keep only real solutions
                real_roots.append((float(sympy.re(approx)), root))
        if not real_roots:
            return ["No real solutions"]

        lines = []
        for approx_value, root in sorted(real_roots, key=lambda pair: pair[0]):
            shown = format_number(approx_value)
            if root.is_Integer:
                lines.append(f"{unknown} = {root}")                       # exact whole number
            elif root.has(sympy.I):
                lines.append(f"{unknown} ≈ {shown}")                      # approximate only
            else:
                exact = str(root).replace("**", "^")
                lines.append(f"{unknown} = {exact} ≈ {shown}")            # exact and approximate
        return lines
```

## 4. interface.py — the window
One shared window for every mode. Flow of a key press:
`key clicked → press() → text goes into the entry box` and `= clicked → calculate() → Calculator.evaluate() → result shown`.

```python
%%writefile interface.py
"""
interface.py - the shared window used by every mode: header, display, mode tabs,
keypad, history / variables panel and keyboard handling.

How a button press flows through the app:
    key clicked -> press(action) -> text goes into the entry box (or a command runs)
    '=' clicked -> calculate() -> calculations.Calculator.evaluate() -> result shown
"""
import re

import customtkinter as ctk

import theme
from calculations import CalcError, Calculator, format_number
from keypads import KEYPADS, MODE_NAMES

OPERATOR_STARTS = "+−×÷%^*"      # keys that continue from the previous answer


class CalcApp(ctk.CTk):
    # ==================== SECTION 1: SET-UP ====================
    def __init__(self):
        super().__init__()
        self.calc = Calculator()           # the maths engine (no GUI inside it)
        self.just_evaluated = False        # True right after '=' was pressed
        self.continue_text = ""            # last numeric answer, used to continue a calculation
        self.panel_visible = False

        self.title("Calc - Scientific & Analytical Calculator")
        self.geometry(theme.WINDOW_SIZE)
        self.minsize(*theme.MIN_SIZE)
        self.configure(fg_color=theme.BG_MAIN)
        self.grid_columnconfigure(0, weight=1)
        self.grid_rowconfigure(3, weight=1)         # the keypad row gets the spare space

        # command keys (the ones starting with ":" in keypads.py) -> methods
        self.commands = {
            ":equals": self.calculate, ":clear": self.clear, ":back": self.back,
            ":mc": self.memory_clear, ":mr": self.memory_recall,
            ":mplus": self.memory_add, ":mminus": self.memory_subtract,
        }

        self.build_header()
        self.build_display()
        self.build_tabs()
        self.keypad = ctk.CTkFrame(self, fg_color="transparent")
        self.keypad.grid(row=3, column=0, sticky="nsew", padx=10, pady=(0, 12))
        self.build_side_panel()
        self.bind_keyboard()

        self.show_keypad("Scientific")
        self.refresh_panels()
        self.entry.focus()

    # ==================== SECTION 2: BUILDING THE WINDOW ====================
    def build_header(self):
        header = ctk.CTkFrame(self, fg_color="transparent")
        header.grid(row=0, column=0, sticky="ew", padx=14, pady=(10, 2))
        header.grid_columnconfigure(0, weight=1)

        titles = ctk.CTkFrame(header, fg_color="transparent")
        titles.grid(row=0, column=0, sticky="w")
        ctk.CTkLabel(titles, text="Calc", font=theme.FONT_TITLE,
                     text_color=theme.TEXT_MAIN).pack(anchor="w")
        ctk.CTkLabel(titles, text="SCIENTIFIC & ANALYTICAL CALCULATOR", font=theme.FONT_SUBTITLE,
                     text_color=theme.TEXT_MUTED).pack(anchor="w")

        self.panel_button = ctk.CTkButton(
            header, text="History", width=84, height=30, command=self.toggle_panel,
            fg_color=theme.TAB_ACTIVE, hover_color=theme.TAB_ACTIVE_HOVER,
            text_color=theme.TEXT_MAIN, font=theme.FONT_PANEL, corner_radius=theme.KEY_RADIUS)
        self.panel_button.grid(row=0, column=1, sticky="e")

    def build_display(self):
        card = ctk.CTkFrame(self, fg_color=theme.BG_DISPLAY, corner_radius=theme.CARD_RADIUS,
                            border_width=1, border_color=theme.BORDER)
        card.grid(row=1, column=0, sticky="ew", padx=14, pady=6)
        card.grid_columnconfigure(0, weight=1)

        # small status row: memory indicator on the left, DEG/RAD switch on the right
        status = ctk.CTkFrame(card, fg_color="transparent")
        status.grid(row=0, column=0, sticky="ew", padx=12, pady=(8, 0))
        status.grid_columnconfigure(1, weight=1)
        self.memory_label = ctk.CTkLabel(status, text="", width=24, font=theme.FONT_STATUS,
                                         text_color=theme.ACCENT)
        self.memory_label.grid(row=0, column=0, sticky="w")
        self.angle_button = ctk.CTkButton(
            status, text="DEG", width=52, height=24, command=self.toggle_angle,
            fg_color=theme.TAB_BAR, hover_color=theme.TAB_ACTIVE, text_color=theme.TEXT_MAIN,
            font=theme.FONT_STATUS, corner_radius=8)
        self.angle_button.grid(row=0, column=2, sticky="e")

        # the expression the user types (also accepts the physical keyboard)
        self.entry = ctk.CTkEntry(card, font=theme.FONT_EXPRESSION, justify="right", height=40,
                                  fg_color=theme.BG_DISPLAY, text_color=theme.TEXT_MAIN,
                                  border_width=0)
        self.entry.grid(row=1, column=0, sticky="ew", padx=10, pady=(2, 0))

        # the result (or a live preview, or an error message)
        self.result_label = ctk.CTkLabel(card, text="0", anchor="e", justify="right", height=84,
                                         wraplength=340, font=theme.FONT_RESULT,
                                         text_color=theme.TEXT_MUTED)
        self.result_label.grid(row=2, column=0, sticky="ew", padx=14, pady=(0, 8))

    def build_tabs(self):
        self.tabs = ctk.CTkSegmentedButton(
            self, values=MODE_NAMES, command=self.show_keypad, height=34, font=theme.FONT_TAB,
            fg_color=theme.TAB_BAR, text_color=theme.TEXT_MAIN,
            selected_color=theme.TAB_ACTIVE, selected_hover_color=theme.TAB_ACTIVE_HOVER,
            unselected_color=theme.TAB_INACTIVE, unselected_hover_color=theme.TAB_INACTIVE_HOVER)
        self.tabs.grid(row=2, column=0, sticky="ew", padx=14, pady=(6, 4))
        self.tabs.set(MODE_NAMES[0])

    def build_side_panel(self):
        """The collapsible panel on the right: history on top, variables below."""
        self.panel = ctk.CTkFrame(self, width=theme.PANEL_WIDTH, fg_color=theme.BG_PANEL,
                                  corner_radius=theme.CARD_RADIUS, border_width=1,
                                  border_color=theme.BORDER)
        self.panel.grid_propagate(False)
        self.panel.grid_columnconfigure(0, weight=1)
        self.panel.grid_rowconfigure(1, weight=3)
        self.panel.grid_rowconfigure(3, weight=2)

        top = ctk.CTkFrame(self.panel, fg_color="transparent")
        top.grid(row=0, column=0, sticky="ew", padx=10, pady=(10, 2))
        top.grid_columnconfigure(0, weight=1)
        ctk.CTkLabel(top, text="History", font=theme.FONT_PANEL_TITLE,
                     text_color=theme.TEXT_MAIN).grid(row=0, column=0, sticky="w")
        ctk.CTkButton(top, text="Clear", width=52, height=24, command=self.clear_history,
                      fg_color=theme.APRICOT, hover_color="#CF8D62", text_color=theme.TEXT_MAIN,
                      font=theme.FONT_PANEL, corner_radius=8).grid(row=0, column=1)

        self.history_list = ctk.CTkScrollableFrame(self.panel, fg_color="transparent")
        self.history_list.grid(row=1, column=0, sticky="nsew", padx=4)

        ctk.CTkLabel(self.panel, text="Variables", font=theme.FONT_PANEL_TITLE,
                     text_color=theme.TEXT_MAIN).grid(row=2, column=0, sticky="w", padx=10, pady=(6, 2))
        self.variable_list = ctk.CTkScrollableFrame(self.panel, fg_color="transparent")
        self.variable_list.grid(row=3, column=0, sticky="nsew", padx=4, pady=(0, 8))

    # ==================== SECTION 3: THE KEYPAD ====================
    def show_keypad(self, mode):
        """Rebuild the keypad for a mode. The expression, history, variables and
        memory live elsewhere, so nothing is lost when the keys change."""
        for widget in self.keypad.winfo_children():
            widget.destroy()
        rows = KEYPADS[mode]
        for column in range(6):
            self.keypad.grid_columnconfigure(column, weight=1, uniform="keys")
        for index in range(12):                       # reset row settings from the last mode
            used = index < len(rows)
            self.keypad.grid_rowconfigure(index, weight=1 if used else 0,
                                          uniform="rows" if used else "")
        for row_number, row in enumerate(rows):
            column = 0
            for key in row:
                label, action, category = key[:3]
                span = key[3] if len(key) == 4 else 1
                background, hover, text_color = theme.KEY_STYLES[category]
                ctk.CTkButton(
                    self.keypad, text=label, width=10, height=10,
                    command=lambda a=action: self.press(a),
                    fg_color=background, hover_color=hover, text_color=text_color,
                    font=theme.FONT_KEY if len(label) <= 2 else theme.FONT_KEY_SMALL,
                    corner_radius=theme.KEY_RADIUS, border_width=1, border_color=theme.KEY_BORDER,
                ).grid(row=row_number, column=column, columnspan=span, sticky="nsew",
                       padx=theme.KEY_GAP, pady=theme.KEY_GAP)
                column += span
        self.tabs.set(mode)
        self.entry.focus()

    def press(self, action):
        """Called by every on-screen key."""
        if action in self.commands:
            self.commands[action]()
        else:
            self.insert_text(action)
        self.entry.focus()

    # ==================== SECTION 4: EDITING THE EXPRESSION ====================
    def set_entry(self, text):
        self.entry.delete(0, "end")
        self.entry.insert(0, text)
        self.entry.icursor("end")

    def insert_text(self, text):
        """Put text into the expression. Right after '=', an operator continues
        from the last answer, anything else starts a new calculation."""
        if self.just_evaluated:
            self.just_evaluated = False
            self.set_entry(self.continue_text if text[0] in OPERATOR_STARTS else "")
        self.entry.insert(self.entry.index("insert"), text)
        self.update_preview()

    def clear(self):
        """C key: clears the current expression only (history, variables, memory stay)."""
        self.set_entry("")
        self.just_evaluated = False
        self.show_result("0", muted=True)

    def back(self):
        """Backspace key: removes a whole function name like 'sin(' in one go."""
        text = self.entry.get()
        shorter = re.sub(r"[A-Za-z]+\($", "", text)
        self.set_entry(shorter if shorter != text else text[:-1])
        self.just_evaluated = False
        self.update_preview()

    # ==================== SECTION 5: CALCULATING AND SHOWING RESULTS ====================
    def calculate(self):
        try:
            result = self.calc.evaluate(self.entry.get())
        except CalcError as error:
            self.show_result(str(error), error=True)
            return
        self.show_result(result)
        self.just_evaluated = True
        try:
            float(result)
            self.continue_text = result              # a number: can be continued
        except ValueError:
            self.continue_text = ""                  # e.g. "x = 2": cannot be continued
        self.refresh_panels()

    def update_preview(self):
        """Show the answer in grey while the user is still typing."""
        if self.just_evaluated:
            return
        preview = self.calc.preview(self.entry.get())
        self.show_result(preview if preview else "0", muted=True)

    def show_result(self, text, muted=False, error=False):
        if error:
            color, font = theme.TEXT_ERROR, theme.FONT_MESSAGE
        else:
            color = theme.TEXT_MUTED if muted else theme.TEXT_MAIN
            if "\n" in text or len(text) > 16:
                font = theme.FONT_RESULT_SMALL
            elif len(text) > 11:
                font = theme.FONT_RESULT_MED
            else:
                font = theme.FONT_RESULT
        self.result_label.configure(text=text, text_color=color, font=font)

    # ==================== SECTION 6: ANGLE MODE AND MEMORY ====================
    def toggle_angle(self):
        self.angle_button.configure(text=self.calc.toggle_angle_mode())
        self.update_preview()
        self.entry.focus()

    def refresh_status(self):
        self.memory_label.configure(text="M" if self.calc.memory else "")

    def memory_clear(self):
        self.calc.memory_clear()
        self.refresh_status()

    def memory_recall(self):
        self.insert_text(self.calc.memory_recall())

    def memory_add(self):
        self._memory_update(self.calc.memory_add)

    def memory_subtract(self):
        self._memory_update(self.calc.memory_subtract)

    def _memory_update(self, operation):
        text = self.entry.get().strip() or "ans"      # use the last answer if the entry is empty
        try:
            operation(text)
        except CalcError as error:
            self.show_result(str(error), error=True)
            return
        self.refresh_status()

    # ==================== SECTION 7: HISTORY AND VARIABLES PANEL ====================
    def toggle_panel(self):
        width, height = self.winfo_width(), self.winfo_height()
        if self.panel_visible:
            self.panel.grid_remove()
            width -= theme.PANEL_WIDTH + 12
        else:
            self.panel.grid(row=0, column=1, rowspan=4, sticky="nsew", padx=(0, 12), pady=12)
            width += theme.PANEL_WIDTH + 12
        self.panel_visible = not self.panel_visible
        self.geometry(f"{width}x{height}")
        self.entry.focus()

    def refresh_panels(self):
        for widget in self.history_list.winfo_children():
            widget.destroy()
        if not self.calc.history:
            ctk.CTkLabel(self.history_list, text="Nothing here yet", font=theme.FONT_PANEL,
                         text_color=theme.TEXT_MUTED).pack(pady=8)
        for expression, result in self.calc.history:
            ctk.CTkButton(
                self.history_list, text=f"{expression}\n= {result}", anchor="w",
                font=theme.FONT_PANEL, fg_color="transparent", hover_color=theme.TAB_INACTIVE_HOVER,
                text_color=theme.TEXT_MAIN, height=44,
                command=lambda e=expression: self.reuse_expression(e),
            ).pack(fill="x", pady=1)

        for widget in self.variable_list.winfo_children():
            widget.destroy()
        if not self.calc.variables:
            ctk.CTkLabel(self.variable_list, text="Try:  a = 5", font=theme.FONT_PANEL,
                         text_color=theme.TEXT_MUTED).pack(pady=8)
        for name, value in sorted(self.calc.variables.items()):
            row = ctk.CTkFrame(self.variable_list, fg_color="transparent")
            row.pack(fill="x")
            ctk.CTkButton(
                row, text=f"{name} = {format_number(value)}", anchor="w", height=28,
                font=theme.FONT_PANEL, fg_color="transparent", hover_color=theme.TAB_INACTIVE_HOVER,
                text_color=theme.TEXT_MAIN, command=lambda n=name: self.insert_text(n),
            ).pack(side="left", fill="x", expand=True)
            ctk.CTkButton(
                row, text="✕", width=28, height=28, font=theme.FONT_PANEL, fg_color="transparent",
                hover_color=theme.APRICOT, text_color=theme.TEXT_MUTED,
                command=lambda n=name: self.remove_variable(n),
            ).pack(side="right")
        self.refresh_status()

    def reuse_expression(self, expression):
        self.set_entry(expression)
        self.just_evaluated = False
        self.update_preview()
        self.entry.focus()

    def remove_variable(self, name):
        self.calc.delete_variable(name)
        self.refresh_panels()

    def clear_history(self):
        self.calc.clear_history()          # only history: expression, variables, memory stay
        self.refresh_panels()

    # ==================== SECTION 8: PHYSICAL KEYBOARD ====================
    def bind_keyboard(self):
        """The entry box already accepts typing. These bindings add Enter, Esc and
        make typing behave the same as the on-screen keys."""
        self.entry.bind("<Return>", lambda event: self.calculate())
        self.entry.bind("<KP_Enter>", lambda event: self.calculate())
        self.entry.bind("<Escape>", lambda event: self.clear())
        self.entry.bind("<Key>", self.on_key_press)
        self.entry.bind("<KeyRelease>", lambda event: self.update_preview())

    def on_key_press(self, event):
        """Runs before a typed character appears in the entry box."""
        if event.keysym in ("BackSpace", "Delete"):
            self.just_evaluated = False
        elif self.just_evaluated and event.char and event.char.isprintable():
            self.just_evaluated = False                   # same rule as insert_text()
            self.set_entry(self.continue_text if event.char in "+-*/^%" else "")
```

## 6. tests.py — automatic checks of the maths

```python
%%writefile tests.py
"""
tests.py - automatic checks for the maths engine and the keypad data.
Run with:  python tests.py
"""
import unittest

import keypads
import theme
from calculations import CalcError, Calculator


def calc(text, angle=None):
    c = Calculator()
    if angle == "RAD":
        c.toggle_angle_mode()
    return c.evaluate(text)


class TestArithmetic(unittest.TestCase):
    def test_basic(self):
        self.assertEqual(calc("5 + 10"), "15")
        self.assertEqual(calc("25 * 4"), "100")
        self.assertEqual(calc("10/4"), "2.5")
        self.assertEqual(calc("7-10"), "-3")

    def test_precedence_and_brackets(self):
        self.assertEqual(calc("2+3*4"), "14")
        self.assertEqual(calc("(2+3)*4"), "20")
        self.assertEqual(calc("-2**2"), "-4")
        self.assertEqual(calc("2(3+4)"), "14")

    def test_display_symbols(self):
        self.assertEqual(calc("2×3"), "6")
        self.assertEqual(calc("9÷3"), "3")
        self.assertEqual(calc("8−3"), "5")
        self.assertEqual(calc("2^3"), "8")

    def test_decimals_and_percent(self):
        self.assertEqual(calc("0.1+0.2"), "0.3")
        self.assertEqual(calc("50%"), "0.5")
        self.assertEqual(calc("2pi"), "6.28318530718")

    def test_auto_close_bracket(self):
        self.assertEqual(calc("sin(30"), "0.5")


class TestScientific(unittest.TestCase):
    def test_roots_powers_logs(self):
        self.assertEqual(calc("sqrt(81)"), "9")
        self.assertEqual(calc("root(27,3)"), "3")
        self.assertEqual(calc("cbrt(-8)"), "-2")
        self.assertEqual(calc("log(1000)"), "3")
        self.assertEqual(calc("ln(e)"), "1")
        self.assertEqual(calc("fact(5)"), "120")
        self.assertEqual(calc("fact(0)"), "1")

    def test_degrees(self):
        self.assertEqual(calc("sin(30)"), "0.5")
        self.assertEqual(calc("cos(60)"), "0.5")
        self.assertEqual(calc("tan(45)"), "1")
        self.assertEqual(calc("asin(0.5)"), "30")
        self.assertEqual(calc("sin(180)"), "0")

    def test_radians(self):
        self.assertEqual(calc("sin(pi/2)", "RAD"), "1")
        self.assertEqual(calc("cos(pi)", "RAD"), "-1")

    def test_advanced(self):
        self.assertEqual(calc("sinh(0)"), "0")
        self.assertEqual(calc("nPr(5,2)"), "20")
        self.assertEqual(calc("nCr(5,2)"), "10")


class TestErrors(unittest.TestCase):
    def assert_error(self, text):
        with self.assertRaises(CalcError):
            Calculator().evaluate(text)

    def test_errors(self):
        for bad in ["1/0", "sqrt(-1)", "log(0)", "tan(90)", "fact(-1)", "fact(2.5)",
                    "fact(171)", "5+", "foo+1", "2**100000", "nPr(1,2,3)", "",
                    "__import__('os')", "(1).real", "'a'", "[1,2]", "1,5", "(-8)**(1/3)"]:
            with self.subTest(bad=bad):
                self.assert_error(bad)


class TestVariablesAndMemory(unittest.TestCase):
    def test_variables(self):
        c = Calculator()
        c.evaluate("a = 5")
        c.evaluate("b = 10")
        self.assertEqual(c.evaluate("a + b"), "15")
        c.evaluate("a = a + 1")
        self.assertEqual(c.evaluate("a"), "6")
        c.delete_variable("a")
        with self.assertRaises(CalcError):
            c.evaluate("a")
        with self.assertRaises(CalcError):
            c.evaluate("sin = 3")

    def test_ans(self):
        c = Calculator()
        c.evaluate("6*7")
        self.assertEqual(c.evaluate("ans + 8"), "50")

    def test_memory(self):
        c = Calculator()
        c.memory_add("5")
        c.memory_add("3")
        self.assertEqual(c.memory_recall(), "8")
        c.memory_subtract("2")
        self.assertEqual(c.memory_recall(), "6")
        c.memory_clear()
        self.assertEqual(c.memory_recall(), "0")

    def test_history(self):
        c = Calculator()
        c.evaluate("1+1")
        c.evaluate("2+2")
        self.assertEqual(c.history[0], ("2+2", "4"))
        c.clear_history()
        self.assertEqual(c.history, [])


class TestSolver(unittest.TestCase):
    def solve(self, text, c=None):
        return (c or Calculator()).evaluate(text).split("\n")

    def test_linear_and_quadratic(self):
        self.assertEqual(self.solve("x + 5 = 12"), ["x = 7"])
        self.assertEqual(self.solve("2x + 3 = 11"), ["x = 4"])
        self.assertEqual(self.solve("x**2 - 4 = 0"), ["x = -2", "x = 2"])
        self.assertEqual(self.solve("x^2 - 5x + 6 = 0"), ["x = 2", "x = 3"])
        self.assertEqual(self.solve("x**3 - 8 = 0"), ["x = 2"])

    def test_exact_and_approximate(self):
        lines = self.solve("x**2 - 2 = 0")
        self.assertTrue(lines[0].startswith("x = -sqrt(2) ≈ -1.4142135623"))
        self.assertEqual(self.solve("2x = 1"), ["x = 1/2 ≈ 0.5"])

    def test_special_cases(self):
        self.assertEqual(self.solve("x**2 + 1 = 0"), ["No real solutions"])
        self.assertEqual(self.solve("x + 1 = x + 2"), ["No solution"])
        self.assertTrue(self.solve("2x + 3 = 3 + 2x")[0].startswith("Every value"))

    def test_uses_variables(self):
        c = Calculator()
        c.evaluate("k = 4")
        self.assertEqual(self.solve("x + k = 10", c), ["x = 6"])

    def test_unsupported(self):
        for bad in ["x + y = 3", "sin(x) = 0", "1/x = 2", "5 = 5", "x**50 = 1", "x = = 2 ="]:
            with self.subTest(bad=bad):
                with self.assertRaises(CalcError):
                    Calculator().evaluate(bad)


class TestKeypadData(unittest.TestCase):
    def test_every_key_is_valid(self):
        for mode, rows in keypads.KEYPADS.items():
            for row in rows:
                width = 0
                for key in row:
                    label, action, category = key[:3]
                    width += key[3] if len(key) == 4 else 1
                    self.assertIn(category, theme.KEY_STYLES, f"{mode}: {label}")
                    if action.startswith(":"):
                        self.assertIn(action, keypads.COMMANDS, f"{mode}: {label}")
                self.assertEqual(width, 6, f"{mode}: a row is not 6 columns wide")


if __name__ == "__main__":
    unittest.main(verbosity=1)
```
